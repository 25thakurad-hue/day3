```python

import csv
with open("emp.csv", "w", newline='') as f:
  w=csv.writer(f)
  w.writerow(["ENO","ENAME","ESALARY","EADDR"])
  n=int(input("enter number of employees: "))
  for i in range(n):
    eno=input("enter employee no: ")
    ename=input("enter employee name: ")
    esal=input("enter employee salary: ")
    eadder=input("enter employee address: ")
    w.writerow([eno, ename, esal, eadder])
print("data inserted sucessfully!")
```

```python
import csv
f=open("emp.csv",'r')
r=csv.reader(f)
data=list(r)
print(data)
for line in data:
  for word in line:
    print(word,"\t",end='')
  print()
```
